### Упражнение 7.2

В ходе выполнения данного упражнения был реализован алгоритм быстрого преобразования Фурье, время работы которого пропорционально $N*logN$. Сначала было проведено тестирование функции FFT из библиотеки numpy

```python
import numpy as np

ys = [-0.5, 0.1, 0.7, -0.1]
hs = np.fft.fft(ys)
print(hs)
```

Теперь выполним дискретное преобразование Фурье и сравним ответы

```python
def dft(ys):
    N = len(ys)
    ts = np.arange(N) / N
    freqs = np.arange(N)
    args = np.outer(ts, freqs)
    M = np.exp(1j * np.pi * 2 * args)
    amps = M.conj().transpose().dot(ys)
    return amps

hs2 = dft(ys)
np.sum(np.abs(hs - hs2))
```

Перед тем как реализовывать FFT рекурсивно, напишем функцию, которая разделяет массив входных данных и применяет FFT к его половинам: 

```python
def fft_halves(ys):
    N = len(ys)
    He = np.fft.fft(ys[::2])
    Ho = np.fft.fft(ys[1::2])
    ns = np.arange(N)
    W = np.exp(-1j * np.pi * 2 * ns / N)
    return np.tile(He, 2) + W * np.tile(Ho, 2)

hs3 = fft_halves(ys)
np.sum(np.abs(hs - hs3))
```

Заменим FFT из nympy на рекурсивные вызовы функции:

```python
def fft(ys):
    N = len(ys)
    if N == 1: return ys
    He = fft(ys[::2])
    Ho = fft(ys[1::2])
    ns = np.arange(N)
    W = np.exp(-1j * np.pi * 2 * ns / N)
    return np.tile(He, 2) + W * np.tile(Ho, 2)
```

```python
hs4 = fft(ys)
np.sum(np.abs(hs - hs4))
```

Получаем результат, близкий к нулю. Таким образом, данная реализация алгоритма быстрого преобразования Фурье является корректной.
